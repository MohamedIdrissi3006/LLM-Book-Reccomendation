```python
import pandas as pd

books = pd.read_csv("book_cleaned.csv")
```

```python
books["categories"].value_counts().reset_index()
```

```python
category_mapping = {
    "Fiction": "Fiction",
    "Juvenile Fiction": "Children's Fiction",
    "Biography & Autobiography": "Nonfiction",
    "History": "Nonfiction",
    "Literary Criticism": "Nonfiction",
    "Philosophy": "Nonfiction",
    "Religion": "Nonfiction",
    "Comics & Graphic Novels": "Fiction",
    "Drama": "Fiction",
    "Juvenile Nonfiction": "Children's Nonfiction",
    "Science": "Nonfiction",
    "Poetry": "Fiction"
}

books["simple_categories"] = books["categories"].map(category_mapping)
```

```python
from transformers import pipeline
```

```python
fiction_categories = ["Fiction","Nonfiction"]
```

```python
pip = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=0
)
```

```python
sequence = books.loc[books["simple_categories"] == "Fiction","description"].reset_index(drop=True)[0]
```

```python
pip(sequence, fiction_categories)
```

```python
import numpy as np

max_index = np.argmax(pip(sequence, fiction_categories)["scores"])
max_label = pip(sequence, fiction_categories)["labels"][max_index]
max_label
```

```python
def generate_prediction(sequence,categories):
  predictions = pip(sequence, fiction_categories)
  max_index = np.argmax(predictions["scores"])
  max_label = predictions["labels"][max_index]
  return max_label
```

```python
from tqdm import tqdm

actual_cats = []
predicted_cats = []
for i in tqdm (range(0, 300)):
  sequence = books.loc[books["simple_categories"] == "Fiction", "description"].reset_index(drop=True)[i]
  predicted_cats += [generate_prediction(sequence, fiction_categories)]
  actual_cats += ["Fiction"]
```

```python
from tqdm import tqdm

actual_cats = []
predicted_cats = []
for i in tqdm (range(0, 300)):
  sequence = books.loc[books["simple_categories"] == "Nonfiction", "description"].reset_index(drop=True)[i]
  predicted_cats += [generate_prediction(sequence, fiction_categories)]
  actual_cats += ["Nonfiction"]
```

```python
predicition_df = pd.DataFrame({"Actual categories":actual_cats,"predicted cats":predicted_cats})
```

```python
predicition_df["correct_prediction"] = (np.where(predicition_df["Actual categories"] == predicition_df["predicted cats"],1,0))
```

```python
predicition_df["correct_prediction"].sum() / len(predicition_df)
```

```python
isbns = []
predicted_cates = []

missing_cats = books.loc[books["simple_categories"].isna(),["isbn13","description"]].reset_index(drop=True)
len(missing_cats)
```

```python
for i in tqdm(range(0,len(missing_cats))):
  sequence = missing_cats["description"][i]
  predicted_cates += [generate_prediction(sequence,fiction_categories)]
  isbns += [missing_cats["isbn13"][i]]
```

```python
print("Number of ISBNs:", len(isbns))
print("Number of predictions:", len(predicted_cates))
print("Difference:", len(isbns) - len(predicted_cates))
missing_predictions = pd.DataFrame({"isbn13":isbns,"predicted_categories":predicted_cates})
```

```python
books = pd.merge(books,missing_predictions,on="isbn13",how="left")
```

```python
books["simple_categories"] = np.where(books["simple_categories"].isna(), books["predicted_categories"], books["simple_categories"])
```

```python
books = books.drop(columns = ["predicted_categories"])
```
